```python
# Definimos la función para consultar
def consultar(sql):
    return pd.read_sql_query(sql, conn)
```

```python
# 2. Selección Simple: Consulta que devuelve solo los nombres de los productos y su precio, ordenados de forma alfabética por nombre.
pd.read_sql_query("""
SELECT producto, precio_unitario FROM ventas_tecnologia
ORDER BY producto ASC
""", conn)
```

```python
# 3. Filtrado Crítico: Obtiene todas las ventas realizadas en el país 'Colombia' donde el precio_unitario es mayor a 500.
pd.read_sql_query("""
SELECT * FROM  ventas_tecnologia
WHERE pais = 'Colombia' AND precio_unitario > 500
""", conn)
```

```python
# 4. Búsqueda de Nulos: Identifica si hay registros donde la categoria sea NULL
pd.read_sql_query("""
SELECT * FROM ventas_tecnologia
WHERE categoria IS NULL
""", conn)
```

```python
# Análisis de Rendimiento (Agregación):Calcula el total de ingresos (cantidad * precio_unitario) por cada categoria. Nombra a la columna resultante como ingresos_totales usando un alias.
pd.read_sql_query("""
SELECT categoria, SUM(cantidad * precio_unitario) AS ingresos_totales FROM ventas_tecnologia
GROUP BY categoria
""", conn)
```

```python
# 6.Filtro de Élite (HAVING): Muestra solo las categorías que han generado más de $10,000 en ingresos totales
pd.read_sql_query("""
SELECT categoria, SUM(cantidad*precio_unitario) AS ingresos_totales FROM ventas_tecnologia
GROUP BY categoria
HAVING ingresos_totales > 10000
""", conn)
```
